```python

def display_menu():
    print("""
=================================
       CRICKET TICKET SYSTEM
=================================

1. Display All Tickets
2. Book a Ticket
3. Display Available Tickets
4. Display Sold Tickets
5. Remaining Tickets
6. Exit
=================================
""")

def display_tickets(tickets):
    print("----------------------------------------")
    print("Seat Number  |  Seat Name   |   Status")
    print("----------------------------------------")
    for ticket in tickets:
        seat_number = ticket[0]
        seat_name = ticket[1]
        status = ticket[2]
        print(f"{seat_number}\t \t{seat_name}\t \t{status}")

def filter_ticket_by_status(status):
    return [ticket for ticket in tickets if ticket[2] == status]

def display_available_tickets():
    available_tickets = filter_ticket_by_status("Available")
    if len(available_tickets) > 0:
        display_tickets(available_tickets)
    else:
        print("*************************************************")
        print("           No Tickets Available to buy           ")
        print("*************************************************")
        
def display_sold_tickets():
    sold_tickets = filter_ticket_by_status("Sold")
    if len(sold_tickets) > 0:
        display_tickets(sold_tickets)
    else:
        print("*************************************************")
        print("     No Tickets Sold Yet! You can buy tickets    ")
        print("*************************************************")
        
def search_ticket_by_seat_number(seat_number):
    for ticket_index, ticket in enumerate(tickets):
        if ticket[0] == seat_number:
            return ticket_index, ticket

def book_ticket():
    seat_number = input("Enter seat number: ")
    print()

    if seat_number.isdigit():
        seat_number = int(seat_number)
        ticket_and_index = search_ticket_by_seat_number(seat_number)
        if ticket_and_index:
            ticket_index, ticket = ticket_and_index
            if ticket[2] != "Sold":
                ticket[2] = "Sold"
                print("Ticket Booked Successfully!")
                print(f"Seat Number: {ticket[0]}")
                print(f"Seat Name: {ticket[1]}")
                print(f"Status: {ticket[2]}")
            else:
                print("***************************************************")
                print(f"     Sorry! This ticket has already been sold     ")
                print("***************************************************")
                
        else:
            print("********************************************************************")
            print(f"     The ticket was not found with seat number '{seat_number}'     ")
            print("********************************************************************")
    else:
        print("*************************************************")
        print(f"        Invalid seat number '{seat_number}'     ")
        print("*************************************************")

def calculate_remaining_tickets():
    available_tickets = filter_ticket_by_status("Available")
    sold_tickets = filter_ticket_by_status("Sold")
    print(f"Total Tickets: {len(tickets)}")
    print(f"Sold Tickets: {len(sold_tickets)}")
    print(f"Available Tickets: {len(available_tickets)}")
    

is_running = True

while is_running:
    display_menu()
    choice = input("Enter your choice (1-6): ")
    print()

    if choice == "1":
        display_tickets(tickets)
    elif choice == "2":
        book_ticket()
    elif choice == "3":
        display_available_tickets()
    elif choice == "4":
        display_sold_tickets()
    elif choice == "5":
        calculate_remaining_tickets() 
    elif choice == "6":
        is_running = False
    else:
        print("You have selected the wrong choice. Please check the menu carefully.")

print("Thanks for using the Ciricket Ticket Booing System 🙏")
```
